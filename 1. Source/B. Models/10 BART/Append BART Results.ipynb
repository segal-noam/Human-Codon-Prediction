## Use best model

```python
test_with_baseline = pd.read_pickle(os.path.join(data_path, '2_test_nn_with_baseline_50.pkl'))

bart_results = pd.read_csv(os.path.join('./Bart30_Mask10to100_TF90to0_Eval100_HumanCodonPrediction', 'bart30_evaluation_2e5_best_masked_decoder_overlap.csv'))
bart_ablation_results = pd.read_csv(os.path.join('./Ablation_2_Bart30_Mask10to100_TF90to0_HumanCodonPrediction_2e5_Eval100NTF', 'ablation2_bart30_evaluation_2e5_best_masked_decoder_overlap.csv'))

print(bart_results['accuracy'].mean())
print(bart_ablation_results['accuracy'].mean())
```

# Merge Dataframes

```python
bart_results['predicted_codons'] = bart_results['predicted_codons'].str.replace(' ', '')

bart_results = bart_results.drop(columns=['inx', 'protein_len', 'protein_AAs', 'protein_codons'])

bart_results.rename(columns={'accuracy': 'bart30_accuracy',
                             'entropy': 'bart30_entropy',
                             'predicted_codons': 'bart30_prediction',
                             'num_of_correct_predicted_codons': 'bart30_num_of_correct_predicted_codons',
                             'perplexity': 'bart30_perplexity',
                             'cross_entropy_loss': 'bart30_cross_entropy_loss'}, inplace=True)


bart_ablation_results['predicted_codons'] = bart_ablation_results['predicted_codons'].str.replace(' ', '')

bart_ablation_results = bart_ablation_results.drop(columns=['inx', 'protein_len', 'protein_AAs', 'protein_codons'])

bart_ablation_results.rename(columns={'accuracy': 'bart30_ablation_accuracy',
                             'entropy': 'bart30_ablation_entropy',
                             'predicted_codons': 'bart30_ablation_prediction',
                             'num_of_correct_predicted_codons': 'bart30_ablation_num_of_correct_predicted_codons',
                             'perplexity': 'bart30_ablation_perplexity',
                             'cross_entropy_loss': 'bart30_ablation_cross_entropy_loss'}, inplace=True)
```

```python
test_with_results = pd.merge(test_with_baseline, bart_results, 
                             right_on='transcript', left_on='query_transcript', how='left')
test_with_results = pd.merge(test_with_results, bart_ablation_results,
                             right_on='transcript', left_on='query_transcript', how='left')
```

# Save Merged Data

**As Pickle**

```python
test_with_results.to_pickle(os.path.join(data_path, '2_test_with_bart30.pkl'))
```

**As Json CSV**

```python
def convert_to_json_string(value):
    if pd.isna(value):
        return value
    return json.dumps(value)

test_with_results['median'] = test_with_results['median'].apply(convert_to_json_string)
test_with_results['median_float'] = test_with_results['median_float'].apply(convert_to_json_string)

test_with_results.to_csv(os.path.join(data_path, '2_test_with_bart30.json.csv'), index=False)
```
