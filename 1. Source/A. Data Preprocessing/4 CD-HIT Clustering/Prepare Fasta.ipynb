# Import Libraries

```python
# Linear algebra
import numpy as np

# Dataframes
import pandas as pd

# Path
import os
data_path = os.path.join('..', 'data')
```

# Load Dataset

```python
# Read discretized expression pickle
csv_dataframe = pd.read_pickle(os.path.join(data_path, 'genes_human_discrete_expression.pkl'))
```

```python
csv_dataframe.info()
```

# Prepare Fasta

```python
# Convert dataframe to fasta file
csv_dataframe['description'] = csv_dataframe.apply(lambda row: f">{row['gene']}|{row['transcript']}", axis=1)
csv_dataframe['fasta_data'] = csv_dataframe.apply(lambda row: f"{row['description']}\n{row['amino_acid_seq']}\n", axis=1)

fasta_data = ''.join(csv_dataframe['fasta_data'])
```

# Save Fasta

```python
with open(os.path.join(data_path, 'proteins_human.fasta'), 'w') as file:
    file.write(fasta_data)
```
